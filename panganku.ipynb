```python
import pandas as pd
import json

# Load JSON file (as list of dicts)
with open('nutrisi_data.json') as f:
    data = json.load(f)

clean_data = [row for row in data if isinstance(row, dict)]

# Convert to DataFrame
df = pd.DataFrame(clean_data)

print(df)
```
